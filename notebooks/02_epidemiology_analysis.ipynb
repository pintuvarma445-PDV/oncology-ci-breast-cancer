```python
import pandas as pd
import matplotlib.pyplot as plt

# Load datasets
world = pd.read_csv("../data/raw/globocan_breast_incidence_world.csv")
india = pd.read_csv("../data/raw/globocan_breast_incidence_india.csv")

world.head()
```

```python
# Top 10 countries by ASR (excluding 'World' row)
world_countries = world[world["Country"] != "World"]
top10_asr = world_countries.sort_values("ASR", ascending=False).head(10)

plt.figure(figsize=(8,5))
plt.barh(top10_asr["Country"], top10_asr["ASR"])
plt.gca().invert_yaxis()
plt.xlabel("Age-standardized incidence rate (per 100,000)")
plt.title("Top 10 Countries – Breast Cancer Incidence (ASR)")
plt.tight_layout()
plt.savefig("../reports/figures/world_top10_asr.png", dpi=300)
plt.show()
```

```python
india_regions = india[india["Region"] != "India Total"]

plt.figure(figsize=(9,5))
plt.bar(india_regions["Region"], india_regions["ASR"])
plt.xticks(rotation=60, ha="right")
plt.ylabel("Age-standardized incidence rate (per 100,000)")
plt.title("Breast Cancer Incidence – Selected India Regions")
plt.tight_layout()
plt.savefig("../reports/figures/india_regions_asr.png", dpi=300)
plt.show()
```
